## Exercise 4: Parse a record

A logger emits records like `"2024-01-15 , JUNGFRAUJOCH , -2.3 , degC"`. Extract the date string, the station name in title case, the numeric value as a float, and the unit as written, then print them. Then print the station name in lower case as well, the form you would use in a file name.

````python
record = "2024-01-15 , JUNGFRAUJOCH , -2.3 , degC"

parts = record.split(",")

date = parts[0].strip()
station_name = parts[1].strip().title()
value = float(parts[2].strip())
unit = parts[3].strip()

print(date, station_name, value, unit)
print(station_name.lower())
````

## Exercise 7: Round-trip through a file, with an append

Write the Jungfraujoch week from the lecture — the daily maximum temperatures -8.4, -8.9, -9.2,
-9.0, -9.5 (°C) — to `_files/jfj_week.txt`, one value per line, using `pathlib` and a `with`
block. Create the `_files` folder first if it does not exist; do not rely on Exercise 6 having
made it.

1. Write the five values with mode `"w"`, one `.write()` call per value.
2. Reopen the file with mode `"a"` and append the last two days of the week, `-7.8` and `-6.1`.
3. Read the file back with `.readlines()`, cast each of the seven lines to `float` by name, and
   compute the mean by summing the seven values by hand with `+` and dividing by `len()`. Print
   it to two decimals.

````python
from pathlib import Path

# Créer le dossier _files s'il n'existe pas
folder = Path("_files")
folder.mkdir(exist_ok=True)

path = folder / "jfj_week.txt"

# Étape 1 : écrire les 5 valeurs, un .write() par valeur
with path.open("w") as f:
    f.write("-8.4\n")
    f.write("-8.9\n")
    f.write("-9.2\n")
    f.write("-9.0\n")
    f.write("-9.5\n")

# Étape 2 : ajouter les deux derniers jours
with path.open("a") as f:
    f.write("-7.8\n")
    f.write("-6.1\n")

# Étape 3 : relire, convertir et calculer la moyenne
with path.open("r") as f:
    lines = f.readlines()

day_1 = float(lines[0])
day_2 = float(lines[1])
day_3 = float(lines[2])
day_4 = float(lines[3])
day_5 = float(lines[4])
day_6 = float(lines[5])
day_7 = float(lines[6])

total = day_1 + day_2 + day_3 + day_4 + day_5 + day_6 + day_7
mean = total / len(lines)

print(f"Mean temperature: {mean:.2f} °C")
````

**Step 1.** Open the file with a `with` block in read mode and read all its lines with
`.readlines()`.

Print how many lines the file has, then print the first three lines exactly as
they come off disk.

````python
from pathlib import Path

with Path(data_file).open("r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()

print(len(lines))
print(lines[0])
print(lines[1])
print(lines[2])
````

**Step 2.** Look at the first record (the second line of the file). Split it on the separator
and print the resulting list. Then print the *type* of each of the three fields.

Answer in a comment: are the temperatures numbers at this point?

````python
first_record = lines[1].strip().split(",")
print(first_record)
print(type(first_record[0]))
print(type(first_record[1]))
print(type(first_record[2]))

# Non, les températures ne sont pas encore des nombres :
# ce sont des chaînes de caractères (str), comme '14.7'.
# Il faudra les convertir avec float() ensuite.
````

**Step 3.** Take that same first record and turn it into three properly typed, well-named
variables: the station code as text, the day as text, and the maximum temperature as a
**float in degrees Celsius**. Print them in one line with an f-string, showing the temperature
to one decimal with its unit.

Recall that °C = (°F − 32) × 5/9.

````python
station_1 = first_record[0]
day_1 = first_record[1]
max_temp_fahrenheit_1 = float(first_record[2])
max_temp_celsius_1 = (max_temp_fahrenheit_1 - 32) * 5 / 9

print(f"{station_1} {day_1} {max_temp_celsius_1:.1f} °C")
````

**Step 4.** Now do the same for the *second* record (the third line of the file), so you have
two days side by side.

1. Ask which of the two days was warmer by comparing the two raw temperature **fields**, before

   wrong.

````python
second_record = lines[2].strip().split(",")
station_2 = second_record[0]
day_2 = second_record[1]
max_temp_fahrenheit_2 = float(second_record[2])
max_temp_celsius_2 = (max_temp_fahrenheit_2 - 32) * 5 / 9

print(f"{station_2} {day_2} {max_temp_celsius_2:.1f} °C")

# 1. Comparaison
print(first_record[2] > second_record[2])    # False

# 2. Comparaison après conversion en float
print(max_temp_fahrenheit_1 > max_temp_fahrenheit_2)   # True

# 3. La bonne réponse est celle avec float : 14.7 °F > 8.6 °F,
#    --> jour 1 était plus chaud.
#    La comparaison de textes est fausse : Python compare les chaînes
#    caractère par caractère, comme dans un dictionnaire (ordre alphabétique).
#    Il compare d'abord '1' et '8' ; comme '1' < '8', il conclut
#    que '14.7' < '8.6', sans tenir compte de la valeur numérique.
````
